```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../../bestsellers_with_categories.csv", index_col="Name")
print(df)
```

```python
df=df[df["Genre"]=="Fiction"]
print(df)
```

```python
plt.bar(df["User Rating"], df["Price"],label="Fiction Books", color="blue")
plt.xlabel("User Rating")
plt.ylabel("Price")
plt.title("User Rating vs Price for Fiction Books")
plt.legend()
plt.show()
```
